```python
from pathlib import Path

from binance_quant_toolkit import audit_klines, load_dataset, performance_summary

muestra = Path("../examples/data/btcusdt_1h_sample.csv")
velas = load_dataset(muestra)
velas.head()
```

## 1. Audita primero

El modelo no debe decidir qué significa una vela ausente. Detecta huecos, duplicados y geometría imposible antes de crear variables.

```python
calidad = audit_klines(velas, interval="1h")
calidad.to_dict()
```

```python
performance_summary(velas, interval="1h")
```
