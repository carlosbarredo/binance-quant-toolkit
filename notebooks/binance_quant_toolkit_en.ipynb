# From market data to a research-ready table

This notebook starts offline. It teaches the data contract, quality gate and descriptive layer. Live downloads are optional. No cell places orders.

```python
from pathlib import Path

from binance_quant_toolkit import audit_klines, load_dataset, performance_summary

sample = Path("../examples/data/btcusdt_1h_sample.csv")
candles = load_dataset(sample)
candles.head()
```

## 1. Audit first

A model should not decide what a missing candle means. Detect gaps, duplicates and impossible price geometry before feature engineering.

```python
quality = audit_klines(candles, interval="1h")
quality.to_dict()
```

## 2. Describe, do not overclaim

The summary uses close-to-close returns. It excludes spread, fees, slippage, funding and latency. It is a data check, not a strategy result.

```python
performance_summary(candles, interval="1h")
```
